## Variables and Data Types

```python
name = "Daksh"
course = "Applied Data Science"
age = 19
fees = 15000.50
completed = True

print("Name :", name)
print("Course :", course)
print("Age :", age)
print("Fees :", fees)
print("Completed :", completed)
```

```python
print(type(name))
print(type(age))
print(type(fees))
print(type(completed))
```

## Operators

```python
a = 25
b = 10

print("Addition =", a + b)
print("Subtraction =", a - b)
print("Multiplication =", a * b)
print("Division =", a / b)
print("Floor Division =", a // b)
print("Modulus =", a % b)
print("Power =", a ** b)
```

```python
print(a > b)
print(a < b)
print(a == b)
print(a != b)
print(a >= b)
print(a <= b)
```

```python
x = True
y = False

print("AND :", x and y)
print("OR :", x or y)
print("NOT :", not x)
```

## Strings

```python
language = "Python"

print("Upper :", language.upper())
print("Lower :", language.lower())
print("Length :", len(language))
print("First Character :", language[0])
print("Last Character :", language[-1])
print("Repeat :", language * 2)
```

## Data Structures

```python
marks = [78, 85, 90, 88, 95]

print("Original List")
print(marks)

marks.append(80)
marks.insert(2, 100)

print("Updated List")
print(marks)

print("Maximum =", max(marks))
print("Minimum =", min(marks))
print("Sum =", sum(marks))
print("Average =", sum(marks)/len(marks))
```

```python
colors = ("Red", "Blue", "Green", "Black")

print(colors)
print("First =", colors[0])
```

```python
numbers = {10,20,30,20,10,40}

print(numbers)

numbers.add(50)
print(numbers)
```

```python
student = {
    "Name":"Deep",
    "Age":20,
    "City":"Surat",
    "Course":"Data Science"
}

print(student)
print(student["Name"])
print(student["Course"])

student["Age"] = 21
print(student)
```

## Control Flow

```python
marks = 82

if marks >= 90:
    print("Grade A")
elif marks >= 75:
    print("Grade B")
elif marks >= 50:
    print("Grade C")
else:
    print("Fail")
```

```python
for i in range(1,6):
    print(i)
```

```python
count = 1

while count <= 5:
    print(count)
    count += 1
```

## Functions

```python
def square(num):
    return num*num

print(square(9))
```

```python
cube = lambda x: x**3

print(cube(4))
```

```python
square_list = [i*i for i in range(1,11)]

print(square_list)
```

## Exception Handling

```python
try:
    num = 10
    result = num / 2
    print(result)
except:
    print("Error")
```
