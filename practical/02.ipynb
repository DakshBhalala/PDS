# Practical 2

**Aim:** Demonstrate creation and manipulation of list, tuple, set, dictionary and string with their built-in methods.

### 1. List

```python
list1 = [10, 20, 30, 20, 40]
list1.append(50)
list1.insert(1, 15)
list1.extend([60, 70])
list1.remove(20)

print("List:", list1)
print("Pop:", list1.pop())
print("Index of 30:", list1.index(30))
print("Count of 20:", list1.count(20))

list1.reverse()
print("Reversed:", list1)

list1.sort()
print("Sorted:", list1)
```

### 2. Tuple

```python
tuple1 = (10, 20, 30, 20, 40)
print("Tuple:", tuple1)
print("Count of 20:", tuple1.count(20))
print("Index of 30:", tuple1.index(30))
print("Length:", len(tuple1))
print("Slice:", tuple1[1:4])
```

### 3. Set

```python
set1 = {1, 2, 3, 4}
set2 = {3, 4, 5, 6}

set1.add(10)
set1.remove(10)

print("Union:", set1.union(set2))
print("Intersection:", set1.intersection(set2))
print("Difference:", set1.difference(set2))
print("Sym Diff:", set1.symmetric_difference(set2))
print("Subset:", {1, 2}.issubset(set1))
print("Disjoint:", set1.isdisjoint({7, 8}))
```

### 4. Dictionary

```python
dict1 = {"name": "Amit", "age": 20, "city": "Surat"}

print("Keys:", list(dict1.keys()))
print("Values:", list(dict1.values()))
print("Items:", list(dict1.items()))
print("Get age:", dict1.get("age"))

dict1.update({"branch": "CE"})
print("Pop city:", dict1.pop("city"))
print("Dict:", dict1)
```

### 5. String

```python
string1 = "  python programming  "

print("Upper:", string1.upper())
print("Lower:", string1.lower())
print("Title:", string1.title())
print("Strip:", string1.strip())
print("Replace:", string1.replace("python", "Java"))
print("Find:", string1.find("prog"))
print("Split:", string1.split())
print("Join:", "-".join(["A", "B", "C"]))
```
