```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import binom, poisson

# Problem:
# A fair coin is tossed 10 times.
# Find the probability of getting exactly 5 heads.
# Then calculate and visualize the Binomial distribution.
n = 10
p = 0.5

probability = binom.pmf(5, n, p)

print("Probability of getting exactly 5 heads:", probability)

# Values of X
x = np.arange(0, 11)

# Binomial probabilities
y = binom.pmf(x, n, p)

print("Binomial Distribution")
for i in range(len(x)):
    print("Number of Heads =", x[i], "Probability =", round(y[i], 4))
```

```python
# Problem:
# A shop receives an average of 4 customers per hour.
# Find the probability of getting exactly 3 customers in one hour.
# Then calculate and visualize the Poisson distribution.
lam = 4

probability = poisson.pmf(3, lam)

print("Probability of getting exactly 3 customers:", probability)

# Values of X
x = np.arange(0, 11)

# Poisson probabilities
y = poisson.pmf(x, lam)
print("Poisson Distribution")
for i in range(len(x)):
    print("Number of Customers =", x[i], "Probability =", round(y[i], 4))
```

```python
plt.figure(figsize=(10, 4))

# Binomial plot
plt.subplot(1, 2, 1)
plt.bar(np.arange(0, 11), binom.pmf(np.arange(0, 11), 10, 0.5), color="Green")
plt.title("Binomial Distribution")
plt.xlabel("Number of Heads")
plt.ylabel("Probability")

# Poisson plot
plt.subplot(1, 2, 2)
plt.bar(np.arange(0, 11), poisson.pmf(np.arange(0, 11), 4), color="red")
plt.title("Poisson Distribution")
plt.xlabel("Number of Customers")
plt.ylabel("Probability")

plt.tight_layout()
plt.show()
```
