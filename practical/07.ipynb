**Aim:** Connect Python with MySQL database and perform Create, Read, Update and Delete (CRUD) operations.

### 1. Database Connection and Table Creation

```python
import mysql.connector

conn = mysql.connector.connect(
    host="localhost",
    user="root"
)
cursor = conn.cursor()

cursor.execute("CREATE DATABASE IF NOT EXISTS company_db")
cursor.execute("USE company_db")
cursor.execute("""
    CREATE TABLE IF NOT EXISTS employees (
        id INT AUTO_INCREMENT PRIMARY KEY,
        name VARCHAR(50),
        department VARCHAR(50),
        salary INT
    )
""")
print("Database and table created successfully.")
```

### 2. Insert Operation (Single Record)

```python
insert_query = "INSERT INTO employees (name, department, salary) VALUES (%s, %s, %s)"
data = ("John Doe", "Engineering", 75000)

cursor.execute(insert_query, data)
conn.commit()
print("Record inserted successfully.")
```

### 3. Select Operation (Read)

```python
cursor.execute("SELECT * FROM employees")
records = cursor.fetchall()

print("Employee records:")
for row in records:
    print(row)
```

### 4. Update Operation (Update)

```python
update_query = "UPDATE employees SET salary = %s WHERE name = %s"
cursor.execute(update_query, (80000, "John Doe"))
conn.commit()
print("Record updated successfully.")

cursor.execute("SELECT * FROM employees")
for row in cursor.fetchall():
    print(row)
```

### 5. Delete Operation (Delete)

```python
delete_query = "DELETE FROM employees WHERE name = %s"
cursor.execute(delete_query, ("John Doe",))
conn.commit()
print("Record deleted successfully.")

cursor.execute("SELECT * FROM employees")
print("Remaining records:", cursor.fetchall())
```

### 6. Multiple Insert Operation (executemany)

```python
insert_many_query = "INSERT INTO employees (name, department, salary) VALUES (%s, %s, %s)"
data_list = [
    ("Alice", "HR", 60000),
    ("Bob", "Finance", 70000),
    ("Charlie", "IT", 80000)
]

cursor.executemany(insert_many_query, data_list)
conn.commit()
print(f"{cursor.rowcount} records inserted successfully.\n")

print("All inserted records:")
cursor.execute("SELECT * FROM employees")
for row in cursor.fetchall():
    print(row)

cursor.close()
conn.close()
print("\nConnection closed.")
```
