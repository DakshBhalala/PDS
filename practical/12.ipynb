**Aim:** Use SciPy's stats module to compute statistical measures and perform statistical tests on datasets.

### 1. Statistical Measures using SciPy

```python
from scipy import stats
import numpy as np

# Create Dataset
data = [10, 20, 20, 30, 40]
print("Dataset:", data)

# Measures of Central Tendency (Mean, Median, Mode)
print("Mean               :", stats.tmean(data))
print("Median             :", np.median(data))
result = stats.mode(data)
print(f"Mode               : {result.mode} (count: {result.count})")

# Measures of Dispersion & Shape (Std Dev, Variance, Skewness, Kurtosis)
print("Standard Deviation :", stats.tstd(data))
print("Variance           :", stats.tvar(data))
print("Skewness           :", round(stats.skew(data), 4))
print("Kurtosis           :", round(stats.kurtosis(data), 4))

# Overall SciPy Describe Summary
print("\nSciPy Describe Summary:\n", stats.describe(data))
```

### 2. Probability Distributions (Normal Distribution)

```python
# Standard Normal Distribution: Probability Density Function (PDF) & Cumulative Distribution Function (CDF)
x = 0
print("Standard Normal PDF at x=0   :", round(stats.norm.pdf(x, loc=0, scale=1), 4))
print("Standard Normal CDF at Z=1.96:", round(stats.norm.cdf(1.96, loc=0, scale=1), 4))
```

### 3. Hypothesis Testing: One-Sample t-Test

```python
# Testing if sample mean equals hypothesized population mean (mu=25)
t_stat, p_val = stats.ttest_1samp(data, popmean=25)

print("Sample:", data)
print("t-statistic:", round(t_stat, 4))
print("p-value    :", round(p_val, 4))

if p_val < 0.05:
    print("Conclusion : Reject Null Hypothesis (Significant difference).")
else:
    print("Conclusion : Fail to Reject Null Hypothesis (No significant difference).")
```

### 4. Correlation Test (Pearson Correlation)

```python
# Pearson Correlation Test
x = [1, 2, 3, 4, 5]
y = [2, 4, 6, 8, 10]

result_corr = stats.pearsonr(x, y)
print("Correlation : ", result_corr.statistic)
print("p-value     : ", result_corr.pvalue)
```
