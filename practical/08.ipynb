**Aim:** Calculate and interpret descriptive statistics: mean, median, mode, variance, standard deviation, range, quartiles and IQR on a given dataset.

### 1. Load Dataset and View Head

```python
import pandas as pd

df = pd.read_csv("employee.csv")
df.head()
```

### 2. Last 5 Records

```python
df.tail(5)
```

### 3. Select Integer Columns

```python
df.select_dtypes("int")
```

### 4. Maximum and Minimum Values

```python
print("Maximum Monthly Income:", df["MonthlyIncome"].max())
print("Minimum Monthly Income:", df["MonthlyIncome"].min())
```

### 5. Measures of Central Tendency

```python
print("Measure of Central Tendency")
print("Mean:", df["MonthlyIncome"].mean())
print("Median:", df["MonthlyIncome"].median())
print("Mode:", df["MonthlyIncome"].mode()[0])
```

### 6. Measures of Spread

```python
print("Measure of Spread")
print("Range:", df["MonthlyIncome"].max() - df["MonthlyIncome"].min())
print("Variance:", df["MonthlyIncome"].var())
print("Standard Deviation:", df["MonthlyIncome"].std())
```

### 7. Quartiles and Interquartile Range (IQR)

```python
# Quartiles
q1 = df["MonthlyIncome"].quantile(0.25)
q2 = df["MonthlyIncome"].quantile(0.50)
q3 = df["MonthlyIncome"].quantile(0.75)
iqr = q3 - q1

print("Q1 (25%):", q1)
print("Q2 (50% / Median):", q2)
print("Q3 (75%):", q3)
print("IQR:", iqr)
```

### 8. Correlation Coefficient

```python
print("Correlation Coefficient")
print("Age vs MonthlyIncome:", df["Age"].corr(df["MonthlyIncome"]))
```
