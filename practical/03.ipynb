**Aim:** Implement programs using if, if-else, elif, while loop, for loop, nested loops, and loop control statements (break, continue, pass).

### 1. Conditional Statements (if, if-else, elif)

```python
# Simple if
n = 10
if n > 0:
    print(f"{n} is positive")

# if-else
if n % 2 == 0:
    print(f"{n} is even")
else:
    print(f"{n} is odd")

# if-elif-else
marks = 75
if marks >= 90:
    print("Grade: A")
elif marks >= 70:
    print("Grade: B")
elif marks >= 50:
    print("Grade: C")
else:
    print("Grade: Fail")
```

### 2. While Loop

```python
i = 1
while i <= 5:
    print(i, end=" ")
    i += 1
print()
```

### 3. For Loop

```python
for i in range(1, 6):
    print(i, end=" ")
print()

colors = ["Red", "Green", "Blue"]
for c in colors:
    print(c)
```

### 4. Nested Loops

```python
for i in range(1, 4):
    for j in range(1, 4):
        print(f"{i}x{j}={i*j}", end="\t")
    print()
```

### 5. Loop Control Statements (break, continue, pass)

```python
print("break example:")
for i in range(1, 6):
    if i == 4:
        break
    print(i, end=" ")
print()

print("continue example:")
for i in range(1, 6):
    if i == 3:
        continue
    print(i, end=" ")
print()

print("pass example:")
for i in range(1, 4):
    if i == 2:
        pass
    print(i, end=" ")
print()
```
