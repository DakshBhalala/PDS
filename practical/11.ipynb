**Aim:** Use Pandas to create Series and DataFrames, perform data selection, filtering, grouping and aggregation operations.

### 1. Creating Series and DataFrame

```python
import pandas as pd

# Pandas Series
s = pd.Series([10, 20, 30, 40, 50], index=["a", "b", "c", "d", "e"])
print("Pandas Series:\n", s)

# Pandas DataFrame
data = {
    "Name": ["Amit", "Pooja", "Rahul", "Neha", "Karan", "Simran"],
    "Department": ["IT", "HR", "IT", "Finance", "HR", "Finance"],
    "Salary": [60000, 50000, 75000, 55000, 52000, 68000],
    "Experience": [3, 2, 5, 3, 2, 4]
}

df = pd.DataFrame(data)
print("\nEmployee DataFrame:\n", df)
```

### 2. Data Selection (loc, iloc, Columns)

```python
# Select single and multiple columns
print("Selecting Name column:\n", df["Name"])
print("\nSelecting Name & Salary:\n", df[["Name", "Salary"]])

# Selection using iloc (position-based)
print("\nFirst 3 rows using iloc:\n", df.iloc[0:3])
print("\nRow 1, Column 0 & 2 using iloc:\n", df.iloc[1, [0, 2]])

# Selection using loc (label-based)
print("\nRows 0 to 2 with specific columns using loc:\n", df.loc[0:2, ["Name", "Department"]])
```

### 3. Data Filtering

```python
# Filter employees with Salary > 55000
print("Employees with Salary > 55000:\n", df[df["Salary"] > 55000])

# Multiple conditions (Department == IT and Experience >= 3)
print("\nIT Department with Exp >= 3:\n", df[(df["Department"] == "IT") & (df["Experience"] >= 3)])
```

### 4. Grouping and Aggregation

```python
# Average salary by department
print("Average Salary by Department:\n", df.groupby("Department")["Salary"].mean())

# Multiple aggregations on Department
print("\nDepartment-wise Aggregation:\n", df.groupby("Department").agg({
    "Salary": ["mean", "max", "min"],
    "Experience": "mean",
    "Name": "count"
}))
```
