**Aim:** Perform read, write and append operations on text files, CSV files, and binary files using Python.

### 1. Text File Operations (Write, Read, Append)

```python
# Write to text file
with open("test.txt", "w") as f:
    f.write("Hello World\nLine 1\n")

# Read from text file
with open("test.txt", "r") as f:
    print("Content after write:")
    print(f.read())

# Append to text file
with open("test.txt", "a") as f:
    f.write("Line 2 (Appended)\n")

# Read after append
with open("test.txt", "r") as f:
    print("Content after append:")
    print(f.read())
```

### 2. CSV File Operations (Write, Read, Append)

```python
import csv

data = [
    ["Roll", "Name", "Marks"],
    [101, "Amit", 85],
    [102, "Pooja", 90]
]

# Write to CSV
with open("test.csv", "w", newline="") as f:
    writer = csv.writer(f)
    writer.writerows(data)

# Read from CSV
print("CSV after write:")
with open("test.csv", "r") as f:
    reader = csv.reader(f)
    for row in reader:
        print(row)

# Append to CSV
with open("test.csv", "a", newline="") as f:
    writer = csv.writer(f)
    writer.writerow([103, "Raj", 78])

# Read after append
print("\nCSV after append:")
with open("test.csv", "r") as f:
    reader = csv.reader(f)
    for row in reader:
        print(row)
```

### 3. Binary File Operations (Write, Read, Append)

```python
# Write to binary file
with open("test.bin", "wb") as f:
    f.write(b"Hello Binary\n")

# Read binary file
with open("test.bin", "rb") as f:
    print("Binary after write:", f.read())

# Append to binary file
with open("test.bin", "ab") as f:
    f.write(b"Appended Binary Line\n")

# Read binary after append
with open("test.bin", "rb") as f:
    print("Binary after append:", f.read())
```
